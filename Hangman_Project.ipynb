```python
import random  # لاستعمال دالة اختيار عنصر عشوائي من القائمة

play_again = "y"  # متغير للتحكم في إعادة تشغيل اللعبة

# تستمر اللعبة طالما اللاعب يختار y
while play_again.lower() == "y":

    # قائمة الكلمات المتاحة
    fruits = ["apple", "mango", "banana", "orange"]

    # اختيار كلمة عشوائية من القائمة
    random_fruits = random.choice(fruits)

    # إنشاء قائمة من الشرطات السفلية بنفس عدد حروف الكلمة
    display = []

    for letter in random_fruits:
        display.append("_")

    # عدد المحاولات المسموح بها
    Attempts = len(random_fruits) + 2

    # عداد الأخطاء لعرض مراحل الرجل المشنوق
    mistakes = 0

    # رسومات الـ Hangman حسب عدد الأخطاء
    hangman = [
    """
     +---+
     |   |
         |
         |
         |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
         |
         |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
     |   |
         |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
    /|   |
         |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
    /|\\  |
         |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
    /|\\  |
    /    |
         |
    =========
    """,
    """
     +---+
     |   |
     O   |
    /|\\  |
    / \\  |
         |
    =========
    """
    ]

    # عرض الكلمة المخفية للاعب
    print(" ".join(display))

    # عرض أول مرحلة من الرسمة
    print(hangman[0])

    # تستمر الجولة طالما يوجد حروف غير مكتشفة ويوجد محاولات متبقية
    while "_" in display and Attempts > 0:

        # إدخال حرف من المستخدم وتحويله لحروف صغيرة
        guessed = input("Please guess a letter: ").lower()

        # خصم محاولة بعد كل إدخال
        Attempts -= 1

        # إذا كان الحرف غير موجود في الكلمة
        if guessed not in random_fruits:

            # زيادة عدد الأخطاء
            mistakes += 1

            print("Wrong guess!")

            # عرض مرحلة الـ Hangman المناسبة لعدد الأخطاء
            print(hangman[min(mistakes, 6)])

        # البحث عن الحرف داخل الكلمة
        for position in range(len(random_fruits)):

            # إذا وُجد الحرف يتم كشف مكانه
            if random_fruits[position] == guessed:
                display[position] = guessed

        # عرض التقدم الحالي في الكلمة
        print(" ".join(display))

        # عرض عدد المحاولات المتبقية
        print(f"You have {Attempts} more tries")

    # إذا لم يتبق أي شرطات فمعنى ذلك أن اللاعب فاز
    if "_" not in display:

        print("""
**********
YOU WIN!
**********
""")

    # إذا انتهت المحاولات قبل تخمين الكلمة
    else:

        print("""
You lose!
""")

        # عرض الشكل النهائي للرجل المشنوق
        print(hangman[6])

        # إظهار الكلمة الصحيحة
        print(f"The word was: {random_fruits}")

    # سؤال المستخدم إذا كان يريد اللعب مرة أخرى
    play_again = input("Play again? (y/n): ")
```
